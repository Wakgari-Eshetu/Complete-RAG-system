RAW FILES
   ↓
LOAD TEXT + METADATA
   ↓
REMOVE PAGE NUMBERS
   ↓
REMOVE HEADERS
   ↓
REMOVE FOOTERS
   ↓
NORMALIZE TEXT
   ↓
REMOVE DUPLICATES
   ↓
PRESERVE STRUCTURE
   ↓
QUALITY CHECK
   ↓
SAVE TO PROCESSED FOLDER

```python
import os

base_folder = "data"

all_documents = []

for root, dirs, files in os.walk(base_folder):
    for file in files:
        if file.endswith(".txt"):
            
            file_path = os.path.join(root, file)

            # read file
            with open(file_path, "r", encoding="utf-8") as f:
                text = f.read()

            # detect disease
            if "malaria" in file_path.lower():
                disease = "malaria"
            elif "diabetes" in file_path.lower():
                disease = "diabetes"
            else:
                disease = "unknown"

            # detect source
            if "who" in file.lower():
                source = "WHO"
            elif "cdc" in file.lower():
                source = "CDC"
            elif "book" in file.lower():
                source = "Textbook"
            else:
                source = "unknown"

            document = {
                "file_path": file_path,
                "file_name": file,
                "disease": disease,
                "source": source,
                "text": text
            }

            all_documents.append(document)

print("Total files loaded:", len(all_documents))
```
